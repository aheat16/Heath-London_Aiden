**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

````python
def powers_of_two(n):
    """Return a list of the powers of 2 from 2^0 through 2^n."""
    powers = []
    for exponent in range(n + 1):
        powers.append(2 ** exponent)
    return powers

# Test the function with the first 10 powers of 2
print(powers_of_two(10))
````

**W.2** Using the function above, plot the first 100 powers of 2

````python
# Plot the first 100 powers of 2: 2^0 through 2^99
powers_100 = powers_of_two(99)

plt.figure(figsize=(10, 6))
plt.plot(range(100), powers_100, marker='o', markersize=3)
plt.xlabel('Exponent')
plt.ylabel('Power of 2')
plt.title('First 100 Powers of 2')
plt.grid(True)
plt.show()
````
